# Installing `SQLite` 
In Ubuntu (WSL)

|#|Method                               | Level         |Pros                      |Cons                                                         |
|-|-|-|-|-|
|1| **APT Package Manager** `APT`       | Beginner      |- Most reliable<br>- easy |- (maybe) Not latest version<br>- Less control of install loc| 
|2| **Download Pre-compiled Binaries**  | Intermediate  |- Latest<br>- portable & fast|- no auto-updates<br>- more work uninstall                   |
|3| **Compile from Source**             | Pro           |- Learning<br>- complete control of compiliation |- no auto-updates <br>- complex|

     

# 1. `apt` Ubuntu Package Manager 
`which sqlite`: Check if already installed

# 1.1 `apt` install
|Command| Desciption|
|-|-|
|`sudo apt update`| apt update|
|`sudo apt install sqlite3 libsqlite3-dev`| install ubuntu packages|
|`sqlite3`| **command-line tool** to interact with SQLite databases.|
|`libsqlite3-dev`| `C` library headers (to compile programs using SQLite).|


# 1.2 `apt` uninstall

|Command| Desciption|
|-|-|
|`sudo apt remove sqlite3 libsqlite3-dev`|  uninstall packages |
|`sudo apt autoremove`|                     remove dependencies|

# 2. Pre-Compiled Binaries
# 2.1 **pre-comp binaries** install
|Step|Command| Desciption|
|-|-|-|
|1|`wget https://www.sqlite.org/2024/sqlite-tools-linux-x64-3450100.zip`    | download precompiled binary|
|2|`unzip sqlite-tools-linux-x64-3450100.zip` & `cd` into unzipped folder   | unzip |
|3|`export PATH=$PATH:$(pwd)`                                               | add to PATH (temporary)|
|3|`echo 'export PATH=$PATH:'$(pwd) >> ~/.bashrc` then source it            | add to PATH permanently by adding to `~/.bashrc`|


# 2.2 **pre-comp binaries** uninstall

|Step|Command| Desciption|
|-|-|-|
|1|`rm -rf sqlite-tools-linux-x64-3450100/` | delete extracted directory|
|2|`vim ~/.bashrc` & remove it              | remove from PATH in ~/.bashrc|

# 3. Compile From Source

# 3.1 **source** install
|#| Step| Code|
|-|-|-|
|1| Extract                          | `tar -xvzf sqlite-autoconf-3500400.tar.gz` & `cd sqlite-autoconf-3500400`|
|2| Configure (can customize prefix) | `./configure --prefix=/usr/local`|
|3| Compile                          | `make`|
|4| Install                          | `sudo make install`|

# 3.2 **source** uninstall

|option|method|code|
|-|-|-|
|1| automated: `make` (if supported)| `sudo make uninstall` |
|2| manually: `rm`  |`sudo rm /usr/local/bin/sqlite3`<br>`sudo rm -r /usr/local/include/sqlite*`<br>`sudo rm /usr/local/lib/libsqlite*`|
